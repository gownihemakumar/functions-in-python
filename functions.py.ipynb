```python
def college_student(name,dept,year):
    print("name:",name)
    print("dept:",dept)
    print("year:",year)
college_student("ravi","Computer science engineering","2nd year")
```

```python
def college_student(name,dept,year,CGPA):
    print("name:",name)
    print("dept:",dept)
    print("year:",year)
    print("CGPA:",CGPA)
college_student(CGPA=8.1,name="ravi",year="2nd year",dept="Computer science engineering")
```

```python
def addition(*args):
    sum=0
    for number in args:
        sum=sum+number
    print(sum)
addition(10,20,30,40,50)
```

```python
def student(**kwargs):
    print(kwargs)
student(name="Ravi", age=19, college="SV University")
```

```python
def student(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)
student(name="Ravi", age=19, branch="CSE")
```
